```python
import pandas as pd
import numpy as np

# Creating the dataset from Exercise 11.4 Question 2
data = {
    'Variable': ['Climate', 'Housing', 'Health', 'Crime', 'Transportation', 'Education', 'Arts', 'Recreation', 'Economy'],
    'PC1': [0.190, 0.544, 0.782, 0.365, 0.585, 0.394, 0.985, 0.520, 0.142],
    'PC2': [0.017, 0.020, -0.605, 0.294, 0.085, -0.273, 0.126, 0.402, 0.150],
    'PC3': [0.207, 0.204, 0.144, 0.585, 0.234, 0.027, 0.111, 0.519, 0.239]
}

df = pd.DataFrame(data)
df.set_index('Variable', inplace=True)
display(df)
```

```python
# Calculate squared loadings to find the variance contribution of each variable per PC
squared_loadings = df ** 2

# Find the variable with the maximum squared loading for each principal component
principal_variables = squared_loadings.idxmax()

print("Principal Variable for each Principal Component (based on max squared loading):")
print(principal_variables)

display(squared_loadings)
```

```python
# Display the original component loadings table from the lab exercise
print("--- Exercise Table: Principal Component Loadings ---")
display(df)

# Find the maximum absolute loading for each variable across the components to see which PC it aligns with most strongly
max_pc_per_variable = df.abs().idxmax(axis=1)
print("\n--- Strongest Principal Component for Each Variable ---")
print(max_pc_per_variable)

# Overall dominant variable (highest loading overall)
dominant_variable = df.abs().stack().idxmax()
print(f"\nThe variable with the overall strongest loading is '{dominant_variable[0]}' on '{dominant_variable[1]}'")
```
